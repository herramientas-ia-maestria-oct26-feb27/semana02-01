## 3. Datos faltantes
No conviene reemplazar automáticamente las notas desconocidas por cero: cambiaríamos el significado de los datos.

```python
mi_df.isna().sum()
```

```python
# Elimina las filas que tienen valores nulos (NaN) en la columna nota_final.
mi_df.dropna(subset=['nota_final'])
```

## 4. Resumir y transformar

```python
mi_df.groupby('carrera')['nota_final'].mean()
```

```python
mi_df.groupby(['carrera', 'modalidad'])['nota_final'].agg(['count', 'mean', 'min', 'max']).round(2)
```

```python
mi_df['nivel_asistencia'] = np.where(mi_df['asistencia'] >= 80, 'Alta', 'Por revisar')
mi_df.head()
```
