## 4. Procesamiento diferido (*lazy*)
`scan_csv()` define una consulta sin materializar de inmediato todos los resultados. `collect()` la ejecuta.

```python
consulta = (pl.scan_csv("data/rendimiento_10000.csv")
    .filter(pl.col("asistencia") >= 80)
    .group_by("carrera")
    .agg(pl.col("nota_final").mean().alias("promedio")))
consulta.collect()
```

## 5. Comparación exploratoria de tiempos
Una medición aislada de 10.000 filas **no demuestra** qué biblioteca es más rápida; incluimos los tiempos para familiarizarnos con la herramienta.

```python
for n in (100, 1000, 10000):
    ruta = f"data/rendimiento_{n}.csv"
    inicio = perf_counter()
    a = pd.read_csv(ruta).groupby("carrera")["nota_final"].mean()
    t_pandas = perf_counter() - inicio
    inicio = perf_counter()
    b = pl.read_csv(ruta).group_by("carrera").agg(pl.col("nota_final").mean())
    t_polars = perf_counter() - inicio
    print(f"{n:>5} registros | Pandas: {t_pandas:.5f}s | Polars: {t_polars:.5f}s")
```
