```python
import pandas as pd
import nltk
from nltk.tokenize import wordpunct_tokenize
from collections import Counter
```

## 1. Cargar comentarios ficticios
Usamos `wordpunct_tokenize` para no exigir descargas de corpus durante la clase.

```python
mi_df = pd.read_csv('data/rendimiento_100.csv')
mi_df[['nota_final', 'comentario']].head()
```

```python
texto = mi_df.loc[0, 'comentario']
print(texto)
```

```python
tokens = wordpunct_tokenize(texto.lower())
tokens
```

## 2. Procesar todos los comentarios

```python
comentarios = mi_df["comentario"].dropna().tolist()
len(comentarios)
```

```python
tokens_todos = []
for comentario in comentarios:
    tokens_todos.extend(wordpunct_tokenize(comentario.lower()))
print(tokens_todos[:40])
```

```python
palabras = [t for t in tokens_todos if t.isalpha()]
Counter(palabras).most_common(15)
```

## 3. Filtrar palabras frecuentes
Esta lista pequeña es demostrativa; no equivale a un corpus completo de palabras vacías.

```python
vacias = {"de", "la", "el", "en", "y", "para", "los", "las", "un", "una", "me", "con", "son", "que"}
frecuentes = Counter(p for p in palabras if p not in vacias)
frecuentes.most_common(15)
```
