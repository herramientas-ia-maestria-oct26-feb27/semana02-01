```python
import pandas as pd
import spacy
from collections import Counter
```

## 1. Tokenizar sin descargar modelos
`spacy.blank("es")` ofrece un tokenizador de español

```python
nlp = spacy.blank("es")
mi_df = pd.read_csv("data/rendimiento_100.csv")
texto = mi_df.loc[0, "comentario"]
texto
```

```python
documento = nlp(texto)
[token.text for token in documento]
```

```python
[(token.text, token.is_punct, token.is_space) for token in documento]
```

## 2. Procesar varias observaciones con `nlp.pipe`

```python
comentarios = mi_df["comentario"].dropna().tolist()
documentos = list(nlp.pipe(comentarios))
len(documentos)
```

```python
palabras = [token.text.lower() for doc in documentos for token in doc if token.is_alpha]
Counter(palabras).most_common(15)
```

## 3. Modelo lingüístico opcional
Con `es_core_news_sm`, podemos explorar categorías gramaticales (`pos_`) y lemas (`lemma_`). La clase funciona también sin esta sección.

```python
try:
    nlp_modelo = spacy.load("es_core_news_sm")
    doc = nlp_modelo(texto)
    print([(t.text, t.lemma_, t.pos_) for t in doc])
except OSError:
    print("Modelo es_core_news_sm no instalado. Continuamos con tokenización básica.")
```
