```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
mi_df = pd.read_csv("data/rendimiento_1000.csv")
mi_df.head()
```

## 1. Histograma de calificaciones

```python
plt.figure(figsize=(8, 4))
plt.hist(mi_df["nota_final"].dropna(), bins=12, edgecolor="black")
plt.title("Distribución de notas finales")
plt.xlabel("Calificación")
plt.ylabel("Número de estudiantes")
plt.show()
```

## 2. Promedio por carrera

```python
promedios = mi_df.groupby("carrera")["nota_final"].mean().sort_values()
promedios
```

```python
plt.figure(figsize=(9, 4))
plt.barh(promedios.index, promedios.values)
plt.title("Promedio de calificaciones por carrera")
plt.xlabel("Calificación promedio")
plt.tight_layout()
plt.show()
```

## 3. Relaciones entre variables

```python
plt.figure(figsize=(8, 4))
plt.scatter(mi_df["horas_estudio"], mi_df["nota_final"], alpha=.35)
# alpha es la transparencia de los puntos
plt.title("Horas de estudio y calificación")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota final")
plt.show()
```

| Horas de estudio | Calificaciones observadas con mayor frecuencia | Interpretación |
|---|---|---|
| **1 a 5 horas** | Entre 40 y 75 puntos | Predominan calificaciones bajas y medias. |
| **8 a 12 horas** | Entre 60 y 90 puntos | Las calificaciones tienden a ser más altas. |
| **15 a 20 horas** | Entre 75 y 100 puntos | Se concentran numerosas calificaciones altas. |

## 4. Dos grupos

```python
resumen = mi_df.groupby("modalidad")["nota_final"].mean()
plt.bar(resumen.index, resumen.values)
plt.title("Promedio por modalidad")
plt.ylabel("Nota final promedio")
plt.show()
```

## 5. Cambiar el tamaño de muestra

```python
df_grande = pd.read_csv("data/rendimiento_10000.csv")
plt.hist(df_grande["nota_final"].dropna(), bins=15, edgecolor="black")
plt.title("Calificaciones en 10.000 registros")
plt.xlabel("Nota final")
plt.ylabel("Frecuencia")
plt.show()
```
